```python
#Q1
data = {    "Name": ["A", "B", "C", "D", "E"],    "Age": [20, 25, 22, 30, 28],    "Marks": [85, 72, 90, 65, 88],    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"] }
df = pd.DataFrame(data)
```

```python
#Q2
data = {    "Age": [20, 21, None, 24, None, 30],    "Salary": [30000, None, 40000, 50000, None, 60000],    "Experience": [1, 2, 3, None, 5, 6],    "Department": ["CS", "IT", "CS", None, "IT", "CS"] }
df = pd.DataFrame(data)
```

```python
#1)
print(df)
#2)
print(df.isnull().sum())
#3)
df['Age'] = df['Age'].fillna(df['Age'].median())
#4)
df['Salary']=df['Salary'].fillna(df['Salary'].median())
#5)
df['Experience']=df['Experience'].fillna(df['Experience'].median())
#6)
df['Department']=df['Department'].fillna(df['Department'].mode())
#7)
print(df.isnull().sum())
#8)
print(df)
```

```python
#Q3
data = {    "Name": ["A", "B", "C", "D", "E", "F"],    "Math": [80, 60, 90, 70, 85, 55],    "Science": [75, 65, 95, 72, 80, 60],    "English": [85, 70, 88, 75, 90, 50] }
df = pd.DataFrame(data)
```

```python
#1)
df['Average']=df[['Math','Science','English']].mean(axis=1)
#2)
df['Total']=df[['Math','Science','English']].sum(axis=1)
#3)
df['Passed']=df['Average']>=60
#4),6)
print(df.sort_values('Average',ascending=False).head())
#5)
print(df[df['Average']>=80])
#7)
print(df['Average'].mean())
#8)
print(df[['Name','Average','Passed']])
```

```python
#Q4
data = {    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H"],    "Department": ["CS", "CS", "IT", "IT", "CS", "IT", "HR", "HR"],    "Salary": [50000, 60000, 55000, 65000, 70000, 75000, 40000, 45000],    "Experience": [2, 4, 3, 5, 6, 7, 1, 2] }
df = pd.DataFrame(data)
```

```python
grouped = df.groupby("Department")

# 1)

print(grouped["Salary"].mean())

# 2)

print(grouped["Salary"].max())

# 3)

print(grouped["Salary"].min())

# 4)
print(grouped["Experience"].mean())

# 5)
avg_salary = grouped["Salary"].mean()
print("Department with highest avg salary:", avg_salary.idxmax(), avg_salary.max())

# 6)

print(grouped["Employee"].count())

# 7)

print(avg_salary.sort_values(ascending=False))
```
